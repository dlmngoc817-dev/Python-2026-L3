```python
#Load both files
table = po.read_csv("students.csv")
df = po.read_csv("scores.csv")
print(table)
print(df)
```

```python
# Check missing values
print(table.isnull().sum())
print(df.isnull().sum())
```

```python
# Fill missing data in students.csv
table["age"] = table["age"].fillna(table["age"].mean()) 
table["GPA"] = table["GPA"].fillna(table["GPA"].mean())
print(table)
```

```python
# Fill missing data in scores.csv
df["math"] = df["math"].fillna(df["math"].mean())
df["database"] = df["database"].fillna(df["database"].mean())
print(df)
```

```python
print(table.isnull().sum())
print(df.isnull().sum())
```

```python
#Merge the two datasets 
import pandas as pd
merged = pd.merge(table, df, on="student_id")
merged
```

```python
# Calculate each student's average score
merged["average_score"] = merged[["python", "math", "database"]].mean(axis=1)

merged
```

```python
# Find the top 5 students
merged.sort_values("average_score", ascending=False).head()
```

```python
# Compute average score by major
merged.groupby("major")["average_score"].mean()
```
